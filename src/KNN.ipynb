```python
import numpy as np
from collections import Counter
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# KNN from scratch - Jupyter cell (can be saved as a .py file)

class KNNClassifier:
    """Simple K-Nearest Neighbors classifier (no optimization)."""
    def __init__(self, k=3, p=2):
        # k: number of neighbors, p: Minkowski p (p=2 -> Euclidean)
        self.k = k
        self.p = p
        self.X_train = None
        self.y_train = None

    def fit(self, X, y):
        """Store training data (lazy learning)."""
        self.X_train = np.array(X)
        self.y_train = np.array(y)

    def _distance(self, a, b):
        """Compute Minkowski distance between two points."""
        return np.linalg.norm(a - b, ord=self.p)

    def _predict_one(self, x):
        """Predict label for a single sample x."""
        # compute distances to all training points
        distances = np.linalg.norm(self.X_train - x, ord=self.p, axis=1)
        # get indices of k nearest neighbors
        nn_idx = np.argsort(distances)[: self.k]
        # majority vote
        nn_labels = self.y_train[nn_idx]
        most_common = Counter(nn_labels).most_common(1)[0][0]
        return most_common

    def predict(self, X):
        """Predict labels for an array of samples."""
        X = np.array(X)
        return np.array([self._predict_one(x) for x in X])

    def score(self, X, y):
        """Compute accuracy on dataset."""
        y_pred = self.predict(X)
        return np.mean(y_pred == np.array(y))


# Demo using Iris dataset
iris = datasets.load_iris()
X, y = iris.data, iris.target

# split into train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# feature scaling helps distance-based methods
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# create, train and evaluate KNN
knn = KNNClassifier(k=5, p=2)
knn.fit(X_train, y_train)

print("Test accuracy:", knn.score(X_test, y_test))

# show a few predictions vs true labels
y_pred = knn.predict(X_test[:10])
print("Pred:", y_pred)
print("True:", y_test[:10])
```
